# Left vs Right Augmented Chow Polynomials for Boolean Matroids $B_n$

This notebook explores and compares the **Right Augmented Chow polynomial** $F(M) = (\underline{H} \cdot f^{\text{rev}})(M)$ and the **Left Augmented Chow polynomial** $G(M) = (g^{\text{rev}} \cdot \underline{H})(M)$ for Boolean matroids $B_n = U_{n, n}$.

## Mathematical Properties of Boolean Matroids:
1. **Isomorphism of Minors:** For any flat $F$ of rank $k$ in $B_n$, the restriction $B_n|_F \cong B_k = U_{k, k}$ and the contraction $B_n / F \cong B_{n-k} = U_{n-k, n-k}$.
2. **Trivial KL Polynomial:** For all Boolean matroids, $P_{B_n}(x) = 1$, so $P^{\text{rev}}_{B_k}(x) = x^k$.
3. **Convolution Identities:**
   $$F(B_n)(x) = \sum_{k=0}^n \binom{n}{k} H(B_k)(x) \cdot x^{n-k}$$
   $$G(B_n)(x) = \sum_{k=0}^n \binom{n}{k} x^k \cdot H(B_{n-k})(x)$$
   By re-indexing $k \to n-k$, it follows immediately that $F(B_n)(x) = G(B_n)(x)$ holds **identically** for all Boolean matroids!
