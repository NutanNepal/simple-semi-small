```sage
P = LaurentPolynomialRing(ZZ, 'x')
x = P.gen()
k, n = 4, 5

M = matroids.CompleteGraphic(n)
#M = matroids.Uniform(k, n)
#M = matroids.catalog.Fano()
E = M.groundset()
L = M.lattice_of_flats()
flats_set = set(L)
r = M.rank()
rank = lambda x : M.rank(x)
mobius = lambda M: characteristic_polynomial(M)(0)
chow = chow_polynomial
redchar = lambda M: - P(reduced_characteristic_polynomial(M)(x))
char_poly = lambda M: P(characteristic_polynomial(M)(x))
kl_poly = lambda M: P(M.lattice_of_flats().kazhdan_lusztig_polynomial()(x))
invkl = lambda M: (-1)**(M.rank()) * P(inverse_kazhdan_lustig_polynomial(M)(x))

print(chow_polynomial(M))
print(reduced_characteristic_polynomial(M))
print(char_poly(M))
print(kl_poly(M))
print(invkl(M))
print(mobius(M))
```

```sage
def sumChow(M):
    flats = list(flats_set)
    flats.remove(E)
    d = {}
    for F in flats:
        d[F] = chow(M.delete(E-F))
    return d, sum(d.values(), P(0))

sumChow(M)[1]
```

```sage
def sumChow2(M):
    flats = list(flats_set)
    flats.remove(frozenset())
    d = {}
    for F in flats:
        d[F] = chow(M.contract(F))
    return d, sum(d.values(), P(0))

sumChow2(M)[1]
```
