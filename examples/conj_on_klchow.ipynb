```sage
P = LaurentPolynomialRing(ZZ, 'x')
x = P.gen()
k, n = 4, 6
M = matroids.CompleteGraphic(n)
#M = matroids.Uniform(k, n)
E = M.groundset()
L = M.lattice_of_flats()
r = M.rank()
rank = lambda x : M.rank(x)
Eq_M = create_deformed_algebra(M, base_ring=P)
E = Eq_M.natural()
E_basis = E.basis()
C = Eq_M.C()
C_basis = C.basis()

Z = Eq_M.zeta()
Z_basis = Z.basis()

flats_keys = list(E_basis.keys())
```

```sage
from IPython.display import HTML, display
import pandas as pd

dict = {F:E_basis[F] for F in flats_keys}
data = []

for F in dict:
    elem = dict[F]
    terms = []

    row_data = {
        'F': str(set(F)),
        'rk(F)': rank(F),
        'in E basis': str(E(elem)),
        'in zeta basis': str(Z(elem)),
        'in C basis': str(C(elem))
    }
    data.append(row_data)

print("Table for E basis")
df = pd.DataFrame(data)
display(HTML(df.to_html(index=True)))
```
