```sage
P = LaurentPolynomialRing(ZZ, 'x')
x = P.gen()
k, n = 5, 7

#M = matroids.CompleteGraphic(n)
M = matroids.Uniform(k, n)
#M = matroids.catalog.Fano()
E = M.groundset()
L = M.lattice_of_flats()
flats_set = set(L)
r = M.rank()
rank = lambda x : M.rank(x)
mobius = lambda M: characteristic_polynomial(M)(0)
chow = chow_polynomial
redchar = lambda M: - reduced_characteristic_polynomial(M)(x)
char_poly = lambda M: characteristic_polynomial(M)(x)
kl_poly = lambda M: M.lattice_of_flats().kazhdan_lusztig_polynomial()(x)
invkl = lambda M: (-1)**(M.rank()) * inverse_kazhdan_lustig_polynomial(M)(x)

print(chow_polynomial(M))
print(reduced_characteristic_polynomial(M))
print(char_poly(M))
print(kl_poly(M))
print(invkl(M))
print(mobius(M))
```

```sage
def augment1(M, f):
    sum = P(0)
    flats = list(flats_set)
    #flats.remove(frozenset())
    for F in flats:
        sum += x**rank(F) * f(M.contract(F))(x)
    return sum

print("augmented chow", augment1(M, chow_polynomial))
print("augmented redchar", augment1(M, redchar))
print("augmented char_poly", augment1(M, char_poly))
print("augmented kl", augment1(M, kl_poly))
print("augmented invkl", augment1(M, invkl))
```

```sage
def augment2(M, f):
    sum = P(0)
    flats = list(flats_set)
    #flats.remove(E)
    for F in flats:
        sum += x**(r - rank(F)) * mobius(M.contract(F)) * f(M.delete(E - F))(x)
    return sum

print("augmented chow", augment2(M, chow_polynomial))
print("augmented redchar", augment2(M, redchar))
print("augmented char_poly", augment2(M, char_poly))
print("augmented kl", augment2(M, kl_poly))
print("augmented invkl", augment2(M, invkl))
```
