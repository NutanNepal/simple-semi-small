```python
P = PolynomialRing(ZZ, 'x')
x = P.gen()
k, n = 5, 5

M = matroids.CompleteGraphic(n)
#M = matroids.Uniform(k, n)
#M = matroids.catalog.Fano()
E = M.groundset()
L = M.lattice_of_flats()
flats_set = set(L)
r = M.rank()
rank = lambda x : M.rank(x)
mobius = lambda M: characteristic_polynomial(M)(0)
chow = chow_polynomial
redchar = lambda M: reduced_characteristic_polynomial(M)(x)
char_poly = lambda M: characteristic_polynomial(M)(x)
kl_poly = lambda M: M.lattice_of_flats().kazhdan_lusztig_polynomial()(x)
invkl = lambda M: (-1)**(M.rank()) * inverse_kazhdan_lustig_polynomial(M)(x)

polys = [chow, redchar, char_poly, kl_poly, invkl]
klpoly = kl_poly(M)
klpolycoeffs = klpoly.list()
print(klpolycoeffs)
```

```python
print("the kazhdan lusztig polynomial is: ", klpoly)
dict_kl = {}

for F in flats_set:
    con = P(kl_poly(M.contract(F)))
    res = P(kl_poly(M.delete(E - set(F))))
    prod = (con * res).list()
    flag = 1
    for i in range(min(len(prod), len(klpolycoeffs))):
        if prod[i] > klpolycoeffs[i]:
            flag = 0
    if flag == 0:
        print(F)
    dict_kl[F] = (con, res, flag)

from IPython.display import HTML, display
import pandas as pd

data_kl = []
for F in flats_set:
    con, res, flag = dict_kl[F]
    data_kl.append({
        'F': str(set(F)),
        'con': con,
        'res': res,
        'prod': (con * res).list(),
        'klpolycoeffs': klpolycoeffs,
        'flag': flag
    })

df_kl = pd.DataFrame(data_kl)
display(HTML(df_kl.to_html(escape=False)))
```
